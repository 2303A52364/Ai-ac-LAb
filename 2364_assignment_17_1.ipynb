```python
import seaborn as sns
import pandas as pd

# Load the healthexp dataset
df = sns.load_dataset('healthexp')

# Display the first 5 rows of the DataFrame
display(df.head())
```

### Pairplot: Exploring Relationships Between All Numerical Variables
Let's start with a `pairplot` to visualize the pairwise relationships and distributions of all numerical columns in the dataset. This helps in understanding correlations and patterns.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create a pairplot
sns.pairplot(df, hue='Country', palette='viridis')
plt.suptitle('Pairplot of Healthexp Dataset by Country', y=1.02) # Add a title for the entire plot
plt.show()
```

### Regplot: Relationship Between Spending and Life Expectancy
Now, let's use a `regplot` to visualize the linear relationship between `Spending_USD` and `Life_Expectancy`. This plot also shows the confidence interval of the regression line.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.regplot(x='Spending_USD', y='Life_Expectancy', data=df, line_kws={'color': 'red'})
plt.title('Relationship Between Spending_USD and Life_Expectancy')
plt.xlabel('Spending_USD (per capita)')
plt.ylabel('Life_Expectancy (Years)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### Boxplot: Life Expectancy Distribution per Country
To see the distribution of `Life_Expectancy` for each `Country`, a `boxplot` is an excellent choice. It shows the median, quartiles, and potential outliers.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))
sns.boxplot(x='Country', y='Life_Expectancy', data=df, palette='viridis')
plt.title('Life Expectancy Distribution per Country')
plt.xlabel('Country')
plt.ylabel('Life_Expectancy (Years)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Scatterplot: Life Expectancy Over Time
Finally, let's use a `scatterplot` to visualize how `Life_Expectancy` has changed over the `Year` for each country. Adding `hue='Country'` will differentiate the trends for each nation.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))
sns.scatterplot(x='Year', y='Life_Expectancy', hue='Country', data=df, palette='tab10', s=100, alpha=0.8)
plt.title('Life Expectancy Over Time by Country')
plt.xlabel('Year')
plt.ylabel('Life_Expectancy (Years)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(title='Country', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Average Life Expectancy per Country
Let's visualize the average `Life_Expectancy` for each `Country` using a bar chart.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the average Life_Expectancy per Country
avg_life_expectancy_per_country = df.groupby('Country')['Life_Expectancy'].mean().reset_index()

# Create a bar chart
plt.figure(figsize=(12, 7))
sns.barplot(x='Country', y='Life_Expectancy', hue='Country', data=avg_life_expectancy_per_country, palette='viridis', legend=False)
plt.title('Average Life Expectancy per Country')
plt.xlabel('Country')
plt.ylabel('Average Life Expectancy (Years)')
plt.xticks(rotation=45, ha='right') # Rotate country names for better readability
plt.tight_layout()
plt.show()
```
