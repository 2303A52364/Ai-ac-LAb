```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Penguin dataset
penguins_df = sns.load_dataset('penguins')

print("Dataset loaded successfully!")
```

### Dataset Overview

```python
# Display the first 5 rows of the dataset
display(penguins_df.head())

# Get a concise summary of the DataFrame
print("\nDataFrame Info:")
penguins_df.info()

# Get descriptive statistics
print("\nDescriptive Statistics:")
display(penguins_df.describe())
```

### Missing Values Analysis

```python
# Check for missing values
print("\nMissing values per column:")
display(penguins_df.isnull().sum())
```

### Distribution of Numerical Features

```python
numeric_cols = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']

plt.figure(figsize=(15, 10))
for i, col in enumerate(numeric_cols):
    plt.subplot(2, 2, i + 1)
    sns.histplot(penguins_df[col].dropna(), kde=True)
    plt.title(f'Distribution of {col}')
plt.tight_layout()
plt.show()
```

### Relationships Between Features

```python
# Pairplot to visualize relationships between all numerical features
sns.pairplot(penguins_df, hue='species', diag_kind='kde')
plt.suptitle('Pair Plot of Penguin Features by Species', y=1.02) # Adjust title position
plt.show()
```

### Categorical Feature Distributions

```python
categorical_cols = ['species', 'island', 'sex']

plt.figure(figsize=(18, 6))
for i, col in enumerate(categorical_cols):
    plt.subplot(1, 3, i + 1)
    sns.countplot(data=penguins_df, x=col, hue=col, palette='viridis', legend=False)
    plt.title(f'Count of Penguins by {col}')
    plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Box Plots of Numerical Features by Species and Island

```python
numeric_cols = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']

plt.figure(figsize=(18, 12))
for i, col in enumerate(numeric_cols):
    plt.subplot(2, 2, i + 1)
    sns.boxplot(data=penguins_df, x='species', y=col, hue='sex', palette='coolwarm')
    plt.title(f'{col} by Species and Sex')
    plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

plt.figure(figsize=(18, 12))
for i, col in enumerate(numeric_cols):
    plt.subplot(2, 2, i + 1)
    sns.violinplot(data=penguins_df, x='island', y=col, hue='species', palette='muted')
    plt.title(f'{col} by Island and Species')
    plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

### Distribution of Penguin Species

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
ax = sns.countplot(data=penguins_df, x='species', hue='species', palette='viridis', legend=False)
plt.title('Count of Penguins by Species')
plt.xlabel('Species')
plt.ylabel('Number of Penguins')

# Add count labels on top of the bars
for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge')

plt.tight_layout()
plt.show()
```

```python
island_counts = penguins_df['island'].value_counts().reset_index()
island_counts.columns = ['Island', 'Count']
print("Count of penguins per island:")
display(island_counts)

plt.figure(figsize=(8, 6))
sns.barplot(x='Island', y='Count', data=island_counts, hue='Island', palette='viridis', legend=False)
plt.title('Count of Penguins by Island')
plt.xlabel('Island')
plt.ylabel('Number of Penguins')
plt.tight_layout()
plt.show()
```
